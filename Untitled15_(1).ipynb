```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
from sklearn.datasets import load_iris

iris = load_iris()

df = pd.DataFrame(data=iris.data, columns=iris.feature_names)
df['target'] = iris.target
df['species'] = df['target'].map({
    0: iris.target_names[0],
    1: iris.target_names[1],
    2: iris.target_names[2]
})
df.head()
```

```python
print('5 dong dau')
display(df.head())
```

```python
print("\n --thong tin du lieu--")
df.info()
```

```python
sns.pairplot(df.drop("target",axis = 1),hue='species',palette = 'Set1',markers = ["o","s","D"])
plt.suptitle("bieu do du lieu")
plt.show()
```

```python
features = ['sepal length (cm)', 'sepal width (cm)', 'petal length (cm)', 'petal width (cm)']

fig, axes = plt.subplots(2, 2, figsize=(14, 10))
fig.suptitle('Biểu đồ Histogram phân bố các đặc trưng theo loài hoa', fontsize=16)

for idx, feature in enumerate(features):
    row = idx // 2
    col = idx % 2
    sns.histplot(data=df, x=feature, hue='species', kde=True, element='step', stat='density', common_norm=False, palette='Set1', ax=axes[row, col])
    axes[row, col].set_title(f'Phân bố của {feature}')

plt.tight_layout()
plt.show()
```

```python
from sklearn.model_selection import train_test_split
x = iris.data
y = iris.target
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=42, stratify=y)

print(f"kich thuoc tap train (X_train) :{x_train.shape}")
print(f"kich thuoc tap test (X_test) :{x_test.shape}")
```

```python
from sklearn.naive_bayes import GaussianNB

model = GaussianNB()
model.fit(x_train, y_train)
print("da huan luyen mo hinh naive bayes")
```

```python
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix

y_pred = model.predict(x_test)
acc = accuracy_score(y_test, y_pred)
print(f"do chinh xac(accuracy) tren tap test : {acc*100:.2f}%\n")
```

```python
cm = confusion_matrix(y_test, y_pred)

plt.figure(figsize=(6, 4))
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues',
            xticklabels=iris.target_names,
            yticklabels=iris.target_names)
plt.xlabel('Dự đoán (Predicted)')
plt.ylabel('Thực tế (Actual)')
plt.title('Ma trận nhầm lẫn (Confusion Matrix)')
plt.show()
```

```python
sample = np.array([[5.1, 3.5, 1.4, 0.2]])

pred_class = model.predict(sample)[0]
pred_species = iris.target_names[pred_class]

probabilities = model.predict_proba(sample)[0]

print(f"mau hoa nhap vao thuoc loai : >>> {pred_species}")
for name, p in zip(iris.target_names, probabilities):
    print(f"  {name}: {p*100:.2f}%")
```

```python
%%writefile part1.py
import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import streamlit as st
from sklearn.datasets import load_iris
from sklearn.metrics import accuracy_score
from sklearn.model_selection import train_test_split
from sklearn.naive_bayes import GaussianNB

st.set_page_config(page_title="Demo Phân Loại Hoa Iris", page_icon="🌸", layout="wide")

species_images = {
    "setosa": "https://upload.wikimedia.org/wikipedia/commons/thumb/5/56/Kosaciec_szczecinkowaty_Iris_setosa.jpg/320px-Kosaciec_szczecinkowaty_Iris_setosa.jpg",
    "versicolor": "https://upload.wikimedia.org/wikipedia/commons/thumb/4/41/Iris_versicolor_3.jpg/320px-Iris_versicolor_3.jpg",
    "virginica": "https://upload.wikimedia.org/wikipedia/commons/thumb/9/9f/Iris_virginica.jpg/320px-Iris_virginica.jpg",
}


@st.cache_resource
def load_model():
    iris = load_iris()
    x, y = iris.data, iris.target

    model = GaussianNB()
    model.fit(x, y)

    x_tr, x_te, y_tr, y_te = train_test_split(
        x, y, test_size=0.2, random_state=42, stratify=y
    )
    acc = accuracy_score(y_te, GaussianNB().fit(x_tr, y_tr).predict(x_te))
    return model, iris, acc


model, iris, test_acc = load_model()
```

```python
%%writefile part2.py


def predict_iris(sepal_length, sepal_width, petal_length, petal_width):
    """Nhận 4 số đo, trả về (tên loài, xác suất của 3 loài)."""
    input_data = np.array([[sepal_length, sepal_width, petal_length, petal_width]])
    prediction = model.predict(input_data)[0]
    probabilities = model.predict_proba(input_data)[0]
    return iris.target_names[prediction], probabilities


def show_image(species):
    """Hiển thị ảnh đúng loài; nếu thiếu ảnh/lỗi mạng thì báo cảnh báo, không làm app bị dừng."""
    local = os.path.join("images", f"{species}.jpg")
    try:
        st.image(local if os.path.exists(local) else species_images[species],
                 caption=f"Iris {species}", width=320)
    except Exception:
        st.warning("Không tải được ảnh minh họa (kiểm tra mạng hoặc thư mục images/).")
```

```python
%%writefile part3.py


with st.sidebar:
    st.header("ℹ️ Thông tin mô hình")
    st.write("**Mô hình:** Gaussian Naive Bayes")
    st.write("**Dữ liệu:** Iris (150 mẫu, 4 đặc trưng, 3 loài)")
    st.metric("Độ chính xác trên tập Test (80/20)", f"{test_acc * 100:.2f}%")

st.title("🌸 Demo Phân Loại Hoa Iris Dataset")
st.markdown("**Mô hình:** Gaussian Naive Bayes")
st.caption("Hướng dẫn: kéo 4 thanh trượt bên trái để chọn kích thước hoa, "
           "sau đó bấm **Dự đoán ngay** để xem kết quả.")
```

```python
%%writefile part4.py


col_left, col_right = st.columns(2)

with col_left:
    st.subheader("Nhập thông tin mẫu hoa")
    sepal_length = st.slider("Chiều dài lá đài (Sepal Length - cm)", 4.0, 8.0, 5.1, 0.1)
    sepal_width = st.slider("Chiều rộng lá đài (Sepal Width - cm)", 2.0, 4.5, 3.5, 0.1)
    petal_length = st.slider("Chiều dài cánh hoa (Petal Length - cm)", 1.0, 7.0, 1.4, 0.1)
    petal_width = st.slider("Chiều rộng cánh hoa (Petal Width - cm)", 0.1, 2.5, 0.2, 0.1)
    clicked = st.button("Dự đoán ngay", type="primary")

with col_right:
    st.subheader("Kết quả dự đoán")
    if clicked:
        species, probs = predict_iris(sepal_length, sepal_width, petal_length, petal_width)
        st.markdown(f"### Loài hoa được dự đoán: **{species.upper()}**")
        show_image(species)

        st.markdown("**Xác suất phân loại:**")
        for name, p in zip(iris.target_names, probs):
            st.write(f"{name.capitalize()}: {p * 100:.2f}%")
            st.progress(float(p))
    else:
        st.info("Chưa có kết quả. Hãy chọn giá trị và bấm **Dự đoán ngay**.")
```

```python
%%writefile part5.py


df = pd.DataFrame(iris.data, columns=iris.feature_names)
df["species"] = [iris.target_names[i] for i in iris.target]

with st.expander("📊 Xem bảng dữ liệu Iris"):
    st.dataframe(df)

with st.expander("📈 Biểu đồ phân bố (Petal Length - Petal Width)"):
    fig, ax = plt.subplots(figsize=(6, 4))
    for name, color in zip(iris.target_names, ["tab:red", "tab:green", "tab:blue"]):
        sub = df[df["species"] == name]
        ax.scatter(sub["petal length (cm)"], sub["petal width (cm)"],
                   label=name, color=color, alpha=0.6)
    ax.scatter(petal_length, petal_width, marker="*", s=300, color="black", label="Điểm bạn nhập")
    ax.set_xlabel("Petal Length (cm)")
    ax.set_ylabel("Petal Width (cm)")
    ax.legend()
    st.pyplot(fig)
```
